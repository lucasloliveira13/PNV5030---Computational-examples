### O que significa cada coluna

Notação: $L$ = comprimento na linha d'água, $B$ = boca (largura máxima na linha d'água), $T$ = calado,
$\nabla$ = volume deslocado, $A_M$ = área da seção mestra (seção transversal submersa de maior área),
$V$ = velocidade, $g$ = aceleração da gravidade.

| Coluna | Símbolo / definição | Faixa no dataset | Significado físico |
|---|---|---|---|
| `center_of_buoyancy` | $LCB$ (% de $L$, em relação à meia-nau; negativo = à ré) | −5.0 a 0.0 | Posição longitudinal do centro de carena (centroide do volume submerso). Indica como o volume está distribuído entre proa e popa; afeta o trim e o padrão de ondas geradas. |
| `prismatic_coefficient` | $C_P = \dfrac{\nabla}{A_M \, L}$ | 0.53 a 0.60 | Quão "cheias" são as extremidades do casco. $C_P$ alto = proa e popa cheias (mais volume útil, em geral mais resistência de ondas em alta velocidade); $C_P$ baixo = extremidades finas. |
| `length_displacement` | $\dfrac{L}{\nabla^{1/3}}$ | 4.34 a 5.14 | Esbeltez do casco: comprimento em relação ao "tamanho" do volume deslocado. Valores altos = casco longo e leve para seu comprimento, típico de menor resistência por unidade de peso. |
| `beam_draught_ratio` | $\dfrac{B}{T}$ | 2.81 a 5.35 | Razão boca/calado. Alto = casco largo e raso (mais estabilidade de forma, maior área molhada relativa); baixo = casco estreito e profundo. |
| `length_beam_ratio` | $\dfrac{L}{B}$ | 2.73 a 3.64 | Razão comprimento/boca. Alto = casco esguio (menos resistência de ondas, menos estabilidade inicial); baixo = casco "bojudo". |
| `froude_number` | $Fn = \dfrac{V}{\sqrt{g\,L}}$ | 0.125 a 0.450 | Velocidade adimensional. Compara o comprimento do casco com o comprimento da onda que ele gera. Perto de $Fn \approx 0.4$ a onda gerada tem comprimento próximo de $L$ e a resistência de ondas cresce de forma abrupta. |
| `resistance` | $R_R / \Delta$ (resistência residual por peso deslocado, $\Delta = \rho g \nabla$) | 0.01 a 62.4 | **Saída do modelo.** Parcela da resistência que sobra depois de descontar o atrito. |

### Por que a saída é a resistência *residual*?

A resistência total ao avanço costuma ser decomposta (hipótese de Froude) em

$$
R_T = R_F + R_R ,
$$

onde $R_F$ é a resistência de atrito, que depende do número de Reynolds $Re = VL/\nu$ e é bem estimada por
linhas empíricas, por exemplo a linha ITTC-57:

$$
C_F = \frac{0.075}{(\log_{10} Re - 2)^2}, \qquad R_F = \tfrac12 \rho V^2 S\, C_F ,
$$

e $R_R$ é a resistência residual (essencialmente formação de ondas + efeitos de forma), que depende da
geometria e de $Fn$, e **não** tem uma fórmula fechada simples. É exatamente essa parcela "difícil" que
o surrogate vai aprender a partir dos ensaios:

$$
\frac{R_R}{\Delta} \approx f\big(LCB,\; C_P,\; L/\nabla^{1/3},\; B/T,\; L/B,\; Fn\big).
$$

Como todas as entradas são adimensionais, um mesmo modelo vale para cascos geometricamente semelhantes
de tamanhos diferentes (semelhança de Froude).

## 2. Preparar os dados para treino

- **Split treino/teste (80/20):** o modelo só vê os pontos de treino; os de teste medem a capacidade de
  generalização.
- **Padronização** de cada coluna $j$ (entradas e saída), usando média $\bar{x}_j$ e desvio $s_j$
  calculados **apenas no treino**:

$$
\tilde{x}_j = \frac{x_j - \bar{x}_j}{s_j}.
$$

Isso é importante porque as escalas variam muito ($Fn \sim 0.1$–$0.5$, $LCB \sim -5$–$0$,
resistência $\sim 0$–$60$). Na MLP, entradas de escalas parecidas deixam os gradientes bem condicionados;
no RBF, a distância entre pontos $\lVert \mathbf{x}-\mathbf{x}' \rVert$ deixa de ser dominada pela
coluna de maior escala. As predições são depois convertidas de volta para a escala original
($y = \tilde{y}\, s_y + \bar{y}$).

```python
feature_cols = ["center_of_buoyancy", "prismatic_coefficient", "length_displacement",
                "beam_draught_ratio", "length_beam_ratio", "froude_number"]
target_col = "resistance"

X = df[feature_cols].values.astype(np.float32)
y = df[target_col].values.astype(np.float32).reshape(-1, 1)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)

x_scaler = StandardScaler().fit(X_train)
y_scaler = StandardScaler().fit(y_train)

X_train_s = x_scaler.transform(X_train).astype(np.float32)
X_test_s = x_scaler.transform(X_test).astype(np.float32)
y_train_s = y_scaler.transform(y_train).astype(np.float32)
y_test_s = y_scaler.transform(y_test).astype(np.float32)

X_train_t = torch.from_numpy(X_train_s)
y_train_t = torch.from_numpy(y_train_s)
X_test_t = torch.from_numpy(X_test_s)
y_test_t = torch.from_numpy(y_test_s)

print("treino:", X_train_t.shape, " teste:", X_test_t.shape)
```

## 3. Definir a MLP (surrogate model)

Arquitetura pequena de propósito: 6 entradas → 2 camadas ocultas (32, 32) com ativação `tanh` → 1 saída.
Para uma entrada padronizada $\mathbf{x} \in \mathbb{R}^6$, a rede calcula

$$
\mathbf{h}_1 = \tanh(\mathbf{W}_1 \mathbf{x} + \mathbf{b}_1), \qquad
\mathbf{h}_2 = \tanh(\mathbf{W}_2 \mathbf{h}_1 + \mathbf{b}_2), \qquad
\hat{y} = \mathbf{w}_3^{\top} \mathbf{h}_2 + b_3 ,
$$

com $\mathbf{W}_1 \in \mathbb{R}^{32\times 6}$, $\mathbf{W}_2 \in \mathbb{R}^{32\times 32}$,
$\mathbf{w}_3 \in \mathbb{R}^{32}$. O conjunto de parâmetros treináveis é
$\boldsymbol{\theta} = \{\mathbf{W}_1, \mathbf{b}_1, \mathbf{W}_2, \mathbf{b}_2, \mathbf{w}_3, b_3\}$:

$$
(6\cdot 32 + 32) + (32\cdot 32 + 32) + (32 + 1) = 224 + 1056 + 33 = 1313 \text{ parâmetros}.
$$

Cada camada faz uma combinação linear seguida de uma não-linearidade suave ($\tanh$). Empilhar camadas
permite representar a forte não-linearidade da resistência perto de $Fn$ alto. O número de parâmetros é
**fixo**, independentemente de quantos pontos de treino existam: isso é o que caracteriza um modelo
paramétrico.

```python
class SurrogateMLP(nn.Module):
    def __init__(self, n_in=6, n_hidden=32, n_out=1):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(n_in, n_hidden), nn.Tanh(),
            nn.Linear(n_hidden, n_hidden), nn.Tanh(),
            nn.Linear(n_hidden, n_out),
        )

    def forward(self, x):
        return self.net(x)

model = SurrogateMLP()
n_params = sum(p.numel() for p in model.parameters())
print(model)
print("Número de parâmetros:", n_params)
```

## 4. Treinar

Treinar a MLP significa encontrar os pesos que minimizam o erro quadrático médio (MSE) sobre os
$N$ pontos de treino:

$$
\mathcal{L}(\boldsymbol{\theta}) = \frac{1}{N}\sum_{i=1}^{N} \big(\hat{y}(\mathbf{x}_i;\boldsymbol{\theta}) - y_i\big)^2 .
$$

O gradiente $\nabla_{\boldsymbol{\theta}}\mathcal{L}$ é obtido por **backpropagation** (regra da cadeia
aplicada camada a camada, feita automaticamente pelo `loss.backward()` do PyTorch). Os pesos são então
atualizados pelo otimizador **Adam**, uma variante do gradiente descendente

$$
\boldsymbol{\theta}_{k+1} = \boldsymbol{\theta}_k - \eta\, \frac{\hat{\mathbf{m}}_k}{\sqrt{\hat{\mathbf{v}}_k} + \epsilon},
$$

onde $\hat{\mathbf{m}}_k$ e $\hat{\mathbf{v}}_k$ são médias móveis do gradiente e do gradiente ao quadrado,
e $\eta = 10^{-2}$ é a taxa de aprendizado. Cada passo desse laço é uma **época** (aqui usamos o conjunto
de treino inteiro a cada passo). Poucas centenas de épocas bastam; roda em segundos em CPU.

Note a analogia com o resto da disciplina: treinar uma rede **é** um problema de otimização
não-linear sem restrições, resolvido por um método baseado em gradiente.

```python
optimizer = torch.optim.Adam(model.parameters(), lr=1e-2)
loss_fn = nn.MSELoss()

n_epochs = 800
history = []

for epoch in range(n_epochs):
    model.train()
    optimizer.zero_grad()
    pred = model(X_train_t)
    loss = loss_fn(pred, y_train_t)
    loss.backward()
    optimizer.step()
    history.append(loss.item())
    if epoch % 100 == 0 or epoch == n_epochs - 1:
        with torch.no_grad():
            test_loss = loss_fn(model(X_test_t), y_test_t).item()
        print(f"epoch {epoch:4d} | train MSE (padronizado) = {loss.item():.4f} | test MSE = {test_loss:.4f}")
```

```python
plt.figure(figsize=(6,4))
plt.plot(history)
plt.yscale("log")
plt.xlabel("época")
plt.ylabel("MSE de treino (escala padronizada, log)")
plt.title("Curva de treinamento da MLP")
plt.tight_layout()
plt.show()
```

## 5. Avaliar: predicted vs. measured

O gráfico clássico de verificação de um surrogate: se os pontos caem sobre a diagonal, o modelo generaliza
bem para os dados de teste (pontos que a rede não viu durante o treino). Duas métricas resumem o ajuste,
calculadas na escala original:

$$
R^2 = 1 - \frac{\sum_i (y_i - \hat{y}_i)^2}{\sum_i (y_i - \bar{y})^2},
\qquad
\text{MAE} = \frac{1}{N}\sum_i |y_i - \hat{y}_i| .
$$

$R^2 = 1$ significa ajuste perfeito; $R^2 = 0$ significa que o modelo não é melhor do que prever sempre a
média. O MAE está nas mesmas unidades da resistência, o que facilita a interpretação de engenharia.

```python
model.eval()
with torch.no_grad():
    pred_test_s = model(X_test_t).numpy()

pred_test = y_scaler.inverse_transform(pred_test_s)
true_test = y_scaler.inverse_transform(y_test_s)

r2 = 1 - np.sum((true_test - pred_test) ** 2) / np.sum((true_test - true_test.mean()) ** 2)
mae = np.mean(np.abs(true_test - pred_test))
print(f"R^2 (teste) = {r2:.4f}   MAE (teste) = {mae:.3f}")

plt.figure(figsize=(5,5))
lims = [0, max(true_test.max(), pred_test.max()) * 1.05]
plt.plot(lims, lims, "k--", lw=1, label="y = x")
plt.scatter(true_test, pred_test, alpha=0.7, edgecolor="k")
plt.xlabel("Resistência medida")
plt.ylabel("Resistência predita (surrogate)")
plt.title(f"Predicted vs. Measured (R² = {r2:.3f})")
plt.legend()
plt.xlim(lims); plt.ylim(lims)
plt.tight_layout()
plt.show()
```

## 7. Segundo surrogate: funções de base radial (RBF)

### 7.1 Por que um segundo surrogate?

A MLP funcionou bem, mas tem duas características que pesam no contexto de engenharia:

- ela tem **1313 pesos** para ajustar a partir de 246 ensaios, e esses pesos não têm significado físico;
- ela devolve só um número $\hat{y}$, **sem dizer o quanto confia nele**. Na Seção 6 isso não
  apareceu porque ficamos dentro da faixa ensaiada, mas fora dela a MLP responde com a mesma "certeza".

Em projeto naval, cada ponto de dado costuma ser caro (um ensaio em tanque, uma rodada de CFD). Faz sentido
então perguntar: **existe um modelo que use diretamente os ensaios que temos, interpolando entre eles, e
que diga onde está sem informação?** É essa a ideia do surrogate por funções de base radial (RBF).

Vamos construí-lo em três passos:

1. **7.2** a ideia em 1D, com um único casco (resistência só em função de $Fn$);
2. **7.3** a extensão para as 6 entradas;
3. **7.4–7.5** a versão probabilística (Gaussian Process), que acrescenta a incerteza e define como o
   modelo é treinado.

### 7.2 A ideia em 1D: interpolar um casco com "sinos" gaussianos

Pegue o casco da Seção 6 e esqueça a geometria: temos apenas pares $(Fn_i,\; R_i)$, com $i = 1, \dots, N$.
O RBF propõe escrever a curva como uma **soma de funções em forma de sino**, uma centrada em cada ensaio:

$$
f(Fn) = \sum_{i=1}^{N} w_i \, \varphi\left(\left| Fn - Fn_i \right|\right)
$$

onde $\varphi$ é a função de base radial gaussiana:

$$
\varphi(r) = \exp\left(-\frac{r^2}{2\ell^2}\right)
$$

- $\varphi$ depende só da **distância** $r$ ao centro (daí o nome *radial*). Vale 1 em cima do ensaio e
  cai para 0 conforme nos afastamos.
- $\ell$ é a **largura** do sino: diz até que distância um ensaio "influencia" a predição.
- $w_i$ é o **peso** (altura) de cada sino. São os únicos coeficientes a determinar.

**Como achar os pesos?** Exigimos que a curva passe exatamente pelos ensaios, $f(Fn_j) = R_j$ para cada
$j$. Isso dá $N$ equações lineares para $N$ incógnitas:

$$
\sum_{i=1}^{N} w_i \, \varphi\left(\left| Fn_j - Fn_i \right|\right) = R_j , \qquad j = 1, \dots, N
$$

ou, em forma matricial:

$$
\Phi \, \mathbf{w} = \mathbf{R}, \qquad \Phi_{ji} = \varphi\left(\left| Fn_j - Fn_i \right|\right)
\qquad \Longrightarrow \qquad \mathbf{w} = \Phi^{-1} \mathbf{R}
$$

Ou seja: **o "treino" de um RBF básico é resolver um sistema linear**. Não há épocas nem gradiente.

Na célula abaixo usamos 8 dos 14 ensaios para construir o RBF e guardamos os outros 6 para conferir a
interpolação. O gráfico da esquerda mostra cada sino $w_i\varphi_i$ (tracejado) e sua soma (linha
cheia). O da direita e a tabela impressa mostram o efeito da largura $\ell$ sobre o erro nos ensaios
guardados, sobre o tamanho dos pesos e sobre o número de condição de $\Phi$.

```python
# Um único casco (o mesmo da Seção 6): 14 ensaios de resistência x Fn
fn_h = real_points["froude_number"].values
r_h = real_points["resistance"].values

idx_tr = np.array([0, 2, 4, 6, 8, 10, 12, 13])          # 8 ensaios usados para construir o RBF
idx_te = np.setdiff1d(np.arange(len(fn_h)), idx_tr)     # 6 ensaios guardados para conferir
fn_c, r_c = fn_h[idx_tr], r_h[idx_tr]                   # centros dos sinos e valores medidos

def phi(r, ell):
    "Função de base radial gaussiana."
    return np.exp(-r**2 / (2 * ell**2))

def fit_rbf_1d(x_c, y_c, ell):
    "Resolve Phi w = y para os pesos."
    Phi = phi(x_c[:, None] - x_c[None, :], ell)          # matriz N x N
    return np.linalg.solve(Phi, y_c)

def eval_rbf_1d(x, x_c, w, ell):
    "f(x) = soma_i w_i * phi(|x - x_i|)."
    return phi(x[:, None] - x_c[None, :], ell) @ w

fn_fine = np.linspace(fn_h.min(), fn_h.max(), 300)
ell_ref = 0.03
w_ref = fit_rbf_1d(fn_c, r_c, ell_ref)
print("Pesos w_i:", np.round(w_ref, 2))

fig, axes = plt.subplots(1, 2, figsize=(11, 4.2))

ax = axes[0]
for wi, ci in zip(w_ref, fn_c):
    ax.plot(fn_fine, wi * phi(fn_fine - ci, ell_ref), "--", lw=1, color="grey")
ax.plot(fn_fine, eval_rbf_1d(fn_fine, fn_c, w_ref, ell_ref), lw=2, label="soma = f(Fn)")
ax.scatter(fn_c, r_c, color="crimson", zorder=5, label="ensaios usados (centros)")
ax.scatter(fn_h[idx_te], r_h[idx_te], facecolor="white", edgecolor="k", zorder=5,
           label="ensaios guardados")
ax.set_title(f"Cada ensaio contribui com um sino (ℓ = {ell_ref})")
ax.set_xlabel("Número de Froude, Fn"); ax.set_ylabel("Resistência residual")
ax.legend(fontsize=8)

ax = axes[1]
print(f"{'ℓ':>6} {'MAE guardados':>14} {'max |w_i|':>12} {'cond(Φ)':>10}")
for ell in [0.01, 0.03, 0.12, 0.3]:
    w = fit_rbf_1d(fn_c, r_c, ell)
    pred_te = eval_rbf_1d(fn_h[idx_te], fn_c, w, ell)
    mae_te = np.mean(np.abs(pred_te - r_h[idx_te]))
    cond = np.linalg.cond(phi(fn_c[:, None] - fn_c[None, :], ell))
    print(f"{ell:>6} {mae_te:>14.2f} {np.abs(w).max():>12.1e} {cond:>10.1e}")
    if ell < 0.3:
        ax.plot(fn_fine, eval_rbf_1d(fn_fine, fn_c, w, ell), lw=2,
                label=f"ℓ = {ell}  (MAE guardados = {mae_te:.2f}, max|w| = {np.abs(w).max():.0e})")
ax.scatter(fn_c, r_c, color="crimson", zorder=5)
ax.scatter(fn_h[idx_te], r_h[idx_te], facecolor="white", edgecolor="k", zorder=5)
ax.set_title("Efeito da largura ℓ")
ax.set_xlabel("Número de Froude, Fn")
ax.legend(fontsize=8)

plt.tight_layout()
plt.show()
```

**O que observar:**

- Com $\ell$ **pequeno** ($0.01$), cada sino é estreito: a curva passa pelos ensaios usados, mas "afunda"
  entre eles porque nenhum sino alcança a região intermediária. O modelo decorou os pontos e erra nos
  ensaios guardados.
- Aumentando $\ell$, os sinos se sobrepõem e a interpolação melhora. Nestes dados, que são uma curva muito
  suave, sinos largos ($0.12$) acertam os ensaios guardados quase perfeitamente.
- **Mas há um preço**: quanto mais largos, mais parecidos os sinos ficam entre si, as colunas de $\Phi$
  ficam quase iguais e a matriz fica **mal condicionada** (veja $\text{cond}(\Phi)$ na tabela). Os pesos
  passam a ter valores enormes, com sinais alternados que se cancelam (lado esquerdo já mostra isso em
  escala menor). Com $\ell = 0.3$ os pesos chegam a $10^{10}$: o resultado fica extremamente sensível a
  qualquer ruído de medição ou erro de arredondamento.

Ou seja: **a largura $\ell$ é o que de fato precisa ser "aprendido"**, e escolhê-la é um compromisso entre
seguir bem os dados e manter o problema numericamente estável. Os pesos $w_i$ saem de graça do sistema
linear depois que $\ell$ está escolhido. Na Seção 7.4 veremos que o termo de ruído $\sigma_n^2$ ajuda na
estabilidade, e na Seção 7.5 que a verossimilhança faz essa escolha de $\ell$ de forma objetiva.

### 7.3 De 1 para 6 entradas: uma largura por variável

No problema completo, cada ensaio é um ponto $\mathbf{x}_i = (LCB,\; C_P,\; L/\nabla^{1/3},\; B/T,\; L/B,\; Fn)$
no espaço de 6 dimensões (padronizado, como na Seção 2). O sino passa a depender da distância entre dois
cascos nesse espaço. Como as 6 entradas não têm a mesma influência na resistência, usamos uma largura
$\ell_d$ **para cada entrada** $d$:

$$
\varphi(\mathbf{x}, \mathbf{x}_i) = \exp\left(-\frac{1}{2} \sum_{d=1}^{6} \frac{(x_d - x_{i,d})^2}{\ell_d^2}\right)
$$

e o surrogate continua sendo uma soma de sinos, agora um por ensaio no espaço 6D:

$$
f(\mathbf{x}) = \sum_{i=1}^{N} w_i \, \varphi(\mathbf{x}, \mathbf{x}_i)
$$

A interpretação de $\ell_d$ é direta:

- $\ell_d$ **pequeno**: basta mudar um pouco a entrada $d$ para "sair do sino", isto é, a resistência é
  sensível a essa entrada;
- $\ell_d$ **grande**: mesmo variando bastante a entrada $d$, o sino quase não muda, logo essa entrada
  pouco influencia a resistência.

Agora são 6 larguras para escolher, e já não dá para fazer isso "no olho" como em 1D. Precisamos de um
critério objetivo, que é o que a visão probabilística fornece.

### 7.4 A visão probabilística: Gaussian Process

Um **Gaussian Process (GP)** reinterpreta o sino como uma **covariância**: dois cascos parecidos devem ter
resistências parecidas (correlacionadas). O sino da Seção 7.3, multiplicado por uma amplitude $\sigma_f^2$,
vira o **kernel RBF**:

$$
k(\mathbf{x}, \mathbf{x}') = \sigma_f^2 \exp\left(-\frac{1}{2} \sum_{d=1}^{6} \frac{(x_d - x'_d)^2}{\ell_d^2}\right)
$$

Com os $N$ ensaios de treino montamos a matriz de covariância, somando um pequeno ruído $\sigma_n^2$ na
diagonal para representar a incerteza experimental de cada ensaio:

$$
K_{ij} = k(\mathbf{x}_i, \mathbf{x}_j) + \sigma_n^2 \, \delta_{ij}
$$

A **predição média** do GP num casco novo $\mathbf{x}_*$ é:

$$
\mu(\mathbf{x}_*) = \sum_{i=1}^{N} \alpha_i \, k(\mathbf{x}_*, \mathbf{x}_i), \qquad \boldsymbol{\alpha} = K^{-1} \mathbf{y}
$$

Compare com a Seção 7.2: é **exatamente a mesma forma** do interpolador RBF, $\sum_i w_i \varphi_i$ com
$\mathbf{w} = \Phi^{-1}\mathbf{R}$. A única diferença é o termo $\sigma_n^2$ na diagonal, que tem dois efeitos: permite à curva não
passar exatamente por cada ensaio quando os dados têm ruído, e melhora o condicionamento de $K$, evitando
a explosão dos pesos vista em 7.2 para sinos largos.

O que o GP **acrescenta** é a variância da predição:

$$
\sigma^2(\mathbf{x}_*) = k(\mathbf{x}_*, \mathbf{x}_*) - \mathbf{k}_*^{\top} K^{-1} \mathbf{k}_*,
\qquad \mathbf{k}_* = \left[ k(\mathbf{x}_*, \mathbf{x}_1), \dots, k(\mathbf{x}_*, \mathbf{x}_N) \right]^{\top}
$$

- O primeiro termo, $k(\mathbf{x}_*, \mathbf{x}_*) = \sigma_f^2$, é a incerteza "a priori", sem dado algum.
- O segundo termo é o quanto os ensaios próximos de $\mathbf{x}_*$ reduzem essa incerteza.
- Perto dos ensaios, $\sigma$ é pequeno. Longe deles, $\mathbf{k}_* \to \mathbf{0}$ e $\sigma \to \sigma_f$.

### 7.5 Como o RBF é treinado

O modelo tem só 8 hiperparâmetros a aprender:

$$
\boldsymbol{\phi} = \left\{ \sigma_f,\; \ell_1, \dots, \ell_6,\; \sigma_n \right\}
$$

**Passo 1: escolher $\boldsymbol{\phi}$.** Em vez de minimizar o erro nos pontos de treino (que um RBF
interpolador zera para qualquer $\ell$, como vimos em 7.2), o GP maximiza a **log-verossimilhança
marginal**, ou seja, a probabilidade de observar os ensaios $\mathbf{y}$ dado o modelo:

$$
\log p(\mathbf{y} \mid X, \boldsymbol{\phi}) =
\underbrace{-\frac{1}{2} \mathbf{y}^{\top} K^{-1} \mathbf{y}}_{\text{ajuste aos dados}}
\;\underbrace{-\frac{1}{2} \log \left| K \right|}_{\text{penalidade de complexidade}}
\;-\; \frac{N}{2} \log 2\pi
$$

- O termo de **ajuste** penaliza modelos que não conseguem explicar os dados (sinos largos demais ou
  amplitude pequena demais para seguir a curva).
- O termo de **complexidade** penaliza modelos flexíveis demais, como o caso $\ell$ pequeno de 7.2, que
  "explica" qualquer conjunto de pontos decorando cada um.

O máximo equilibra os dois, escolhendo automaticamente as larguras, agora para as 6 entradas ao mesmo
tempo, junto com o nível de ruído $\sigma_n$. A maximização é um problema de otimização não-linear com
limites, resolvido com L-BFGS-B (um método quasi-Newton, da família vista no tópico de programação
não-linear), com alguns reinícios aleatórios para evitar máximos locais.

**Passo 2: calcular os pesos.** Com $\boldsymbol{\phi}$ fixo, resolve-se o sistema linear
$K\boldsymbol{\alpha} = \mathbf{y}$ (por decomposição de Cholesky).

| | MLP | RBF / GP |
|---|---|---|
| O que se ajusta | 1313 pesos | 8 hiperparâmetros + $N$ pesos $\alpha_i$ |
| Como | gradiente (Adam), por épocas | máx. verossimilhança (L-BFGS-B) + sistema linear |
| Custo | linear em $N$ por época | $\mathcal{O}(N^3)$ pela matriz $K$ |

O custo $\mathcal{O}(N^3)$ é trivial para os 246 ensaios de treino, mas inviável para centenas de
milhares de pontos. É a razão de o GP ser típico de problemas com **poucos dados caros**.

Na célula abaixo, `ConstantKernel` é o $\sigma_f^2$, `RBF` contém os $\ell_d$ e `WhiteKernel` é o
$\sigma_n^2$.

```python
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import RBF, ConstantKernel, WhiteKernel

# k(x, x') = sigma_f^2 * exp(-1/2 * sum_d (x_d - x'_d)^2 / l_d^2) + sigma_n^2 * delta
kernel = (ConstantKernel(1.0, (1e-3, 1e3))                                     # sigma_f^2
          * RBF(length_scale=np.ones(len(feature_cols)),                       # l_1 ... l_6
                length_scale_bounds=(1e-2, 1e3))
          + WhiteKernel(noise_level=1e-4, noise_level_bounds=(1e-8, 1e-1)))    # sigma_n^2

gp = GaussianProcessRegressor(kernel=kernel, n_restarts_optimizer=5, random_state=0)
gp.fit(X_train_s, y_train_s.ravel())   # mesmos dados padronizados usados na MLP

print("Kernel antes do treino :", kernel)
print("Kernel depois do treino:", gp.kernel_)
print(f"Log-verossimilhança marginal no ótimo: {gp.log_marginal_likelihood_value_:.2f}")
```

### 7.6 O que o RBF aprendeu: larguras $\ell_d$

Como as entradas estão padronizadas, os $\ell_d$ são comparáveis entre si. Usando $1/\ell_d$ como medida
de relevância, espera-se que $Fn$ apareça como a entrada mais influente, coerente com a Seção 1.

```python
length_scales = gp.kernel_.k1.k2.length_scale
ls_table = pd.DataFrame({"entrada": feature_cols, "largura l_d": length_scales})
ls_table = ls_table.sort_values("largura l_d")
print(ls_table.to_string(index=False, float_format=lambda v: f"{v:.3g}"))

plt.figure(figsize=(6, 3.5))
plt.barh(ls_table["entrada"], 1 / ls_table["largura l_d"])
plt.gca().invert_yaxis()
plt.xlabel("1 / l_d  (maior = entrada mais influente)")
plt.title("Relevância das entradas segundo o kernel RBF")
plt.tight_layout()
plt.show()
```

### 7.7 Comparação MLP × RBF no conjunto de teste

Mesmos pontos de teste e mesmas métricas ($R^2$ e MAE) da Seção 5. No gráfico do RBF, as barras verticais
mostram $\mu \pm 2\sigma$, o intervalo de aproximadamente 95% de confiança que o próprio modelo informa.

```python
mu_test_s, std_test_s = gp.predict(X_test_s, return_std=True)
mu_test = y_scaler.inverse_transform(mu_test_s.reshape(-1, 1))
std_test = std_test_s * y_scaler.scale_[0]          # desvio volta para a escala original

r2_rbf = 1 - np.sum((true_test - mu_test) ** 2) / np.sum((true_test - true_test.mean()) ** 2)
mae_rbf = np.mean(np.abs(true_test - mu_test))

print(f"MLP : R^2 = {r2:.4f}   MAE = {mae:.3f}")
print(f"RBF : R^2 = {r2_rbf:.4f}   MAE = {mae_rbf:.3f}")

fig, axes = plt.subplots(1, 2, figsize=(10, 4.6), sharex=True, sharey=True)
lims = [0, max(true_test.max(), pred_test.max(), mu_test.max()) * 1.05]

axes[0].scatter(true_test, pred_test, alpha=0.7, edgecolor="k")
axes[0].set_title(f"MLP (R² = {r2:.3f})")

axes[1].errorbar(true_test.ravel(), mu_test.ravel(), yerr=2 * std_test,
                 fmt="o", alpha=0.7, mec="k", capsize=2)
axes[1].set_title(f"RBF / GP (R² = {r2_rbf:.3f}), barras = ±2σ")

for ax in axes:
    ax.plot(lims, lims, "k--", lw=1)
    ax.set_xlim(lims); ax.set_ylim(lims)
    ax.set_xlabel("Resistência medida")
axes[0].set_ylabel("Resistência predita")
plt.tight_layout()
plt.show()
```

### 7.8 Curva de resistência vs. Froude: as duas predições lado a lado

Mesmo casco de referência da Seção 6. A faixa sombreada é $\mu \pm 2\sigma$ do RBF. Estendemos o eixo
além da faixa ensaiada ($Fn > 0.45$) de propósito:

- **dentro** da faixa, as duas curvas praticamente coincidem e $\sigma$ é quase nulo;
- **fora** da faixa, as curvas divergem. A MLP extrapola sem nenhum aviso, enquanto o RBF mostra, pelo
  crescimento de $\sigma$, que está sem dados ali. Isso é a fórmula de $\sigma^2(\mathbf{x}_*)$ da Seção
  7.4 em ação.

```python
fn_ext = np.linspace(0.125, 0.55, 150)          # inclui uma extrapolação além de Fn = 0.45
X_ext = np.column_stack([np.tile(base_geom, (len(fn_ext), 1)), fn_ext]).astype(np.float32)
X_ext_s = x_scaler.transform(X_ext).astype(np.float32)

with torch.no_grad():
    mlp_ext = y_scaler.inverse_transform(model(torch.from_numpy(X_ext_s)).numpy()).ravel()
mu_ext_s, std_ext_s = gp.predict(X_ext_s, return_std=True)
mu_ext = y_scaler.inverse_transform(mu_ext_s.reshape(-1, 1)).ravel()
std_ext = std_ext_s * y_scaler.scale_[0]

plt.figure(figsize=(7, 4.5))
plt.plot(fn_ext, mlp_ext, lw=2, label="MLP")
plt.plot(fn_ext, mu_ext, lw=2, label="RBF / GP (média μ)")
plt.fill_between(fn_ext, mu_ext - 2 * std_ext, mu_ext + 2 * std_ext, alpha=0.25, label="RBF μ ± 2σ")
plt.scatter(real_points["froude_number"], real_points["resistance"], color="crimson", zorder=5,
            label="Casco real mais próximo (dados)")
plt.axvspan(0.45, fn_ext.max(), color="grey", alpha=0.12, label="Fora da faixa ensaiada")
plt.xlabel("Número de Froude, Fn")
plt.ylabel("Resistência residual")
plt.title("Resistência vs. Fn: MLP × RBF")
plt.legend(fontsize=8)
plt.tight_layout()
plt.show()
```

### 7.9 Resumo da comparação

| | MLP | RBF / Gaussian Process |
|---|---|---|
| Forma do modelo | camadas $\tanh$ empilhadas (Seção 3) | soma de sinos: $\mu(\mathbf{x}) = \sum_i \alpha_i k(\mathbf{x}, \mathbf{x}_i)$ |
| Tipo | paramétrico (nº de pesos fixo) | não-paramétrico (um peso por ensaio) |
| O que é ajustado | 1313 pesos | 8 hiperparâmetros + $\boldsymbol{\alpha} = K^{-1}\mathbf{y}$ |
| Como | backpropagation + Adam, por épocas | máx. verossimilhança + sistema linear |
| Incerteza | não fornece | fornece $\sigma(\mathbf{x})$ diretamente |
| Interpretação | pesos sem significado físico | $\ell_d$ indica a relevância de cada entrada |
| Quando usar | muitos dados, alta dimensão | poucos dados caros (CFD, ensaio), otimização bayesiana |

Para dados caros, como ensaios em tanque, o RBF costuma ser a escolha natural: com poucas centenas de
pontos ele já ajusta bem e ainda indica **onde** não confia em si mesmo. A MLP ganha quando o volume de
dados é grande ou quando se quer embutir física na função de perda (Aula 2).

## 8. Conectando com otimização (fecho da aula)

Com o surrogate treinado, uma avaliação de `(geometria, Fn) → resistência` custa microssegundos, contra
horas de CFD/ensaio. Isso é o que torna viável, por exemplo, varrer ou otimizar a geometria do casco para
minimizar resistência em uma faixa de operação. O problema de projeto vira

$$
\min_{\mathbf{g}} \; \bar{R}(\mathbf{g}) = \frac{1}{M} \sum_{m=1}^{M} \hat{f}(\mathbf{g}, Fn_m),
\qquad \mathbf{g} = (LCB,\; C_P,\; L/\nabla^{1/3},\; B/T,\; L/B)
$$

em que $\hat{f}$ é o surrogate (MLP ou RBF) e a média é tomada ao longo da faixa de velocidades de
operação. O mesmo surrogate pode ser a função objetivo de grid search, gradiente (MLP e RBF são
diferenciáveis) ou algoritmo evolutivo. Com o RBF, dá ainda para usar $\sigma$ para decidir onde vale a
pena fazer o próximo ensaio caro (otimização bayesiana). Abaixo, uma versão simples: avaliar as 22
geometrias ensaiadas com a MLP.

```python
# Ilustração simples: qual das 22 geometrias do dataset minimiza a resistência média
# ao longo de toda a faixa de Fn, segundo o surrogate treinado?
geoms = df[feature_cols[:-1]].drop_duplicates().values

best_geom, best_mean_r = None, np.inf
results = []
for g in geoms:
    Xg = np.column_stack([np.tile(g, (len(fn_grid), 1)), fn_grid]).astype(np.float32)
    Xg_s = x_scaler.transform(Xg).astype(np.float32)
    with torch.no_grad():
        pg_s = model(torch.from_numpy(Xg_s)).numpy()
    pg = y_scaler.inverse_transform(pg_s)
    mean_r = pg.mean()
    results.append(mean_r)
    if mean_r < best_mean_r:
        best_mean_r = mean_r
        best_geom = g

print("Resistência média (surrogate) mais baixa entre as 22 geometrias:", round(best_mean_r, 3))
print("Geometria correspondente (center_of_buoyancy, Cp, L/D, B/Dr, L/B):", np.round(best_geom, 3))
```
